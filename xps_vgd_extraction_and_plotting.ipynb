# Start-to-end .VGD extraction workflow:
1. import packages
2. read .VGD file
3. convert spectra to pandas DataFrame
4. export to CSV
5. plot normalised intensity

## Setup

```python
from pathlib import Path
import os
import sys

XPS_DIR = Path(
    "/Users/liangze/Desktop/Tsinghua 2026 Summer/"
    "pdi_h2o2_production/data/XPS"
)

os.chdir(XPS_DIR)

if str(XPS_DIR) not in sys.path:
    sys.path.insert(0, str(XPS_DIR))

print("Working directory:", Path.cwd())
print("Python:", sys.executable)
```

```python
from xps_vgd_utils import (
    read_vgd_to_dataframe,
    add_normalised_intensity,
    save_xps_outputs,
    save_khervefitting_xlsx,
    extract_peak_maximum,
    plot_normalised_xps,
)

PEAK_WINDOWS = {
    "C 1s": (280, 292),
    "N 1s": (395, 402),
    "O 1s": (526, 536),
    "S 2p": (158, 170),
    "Cl 2p": (195, 205),
    "Zn 2p3/2": (1018, 1026),
    "In 3d5/2": (441, 448),
}

CORE_LEVEL_COLOURS = {
    # Canonical labels used in this notebook.
    "C 1s": "#8C8C8C",      # carbon: neutral grey
    "N 1s": "#2F80ED",      # nitrogen: blue
    "O 1s": "#EB5757",      # oxygen: red
    "S 2p": "#F2C94C",      # sulfur: gold/yellow
    "Cl 2p": "#27AE60",     # chlorine: green

    # Aliases commonly parsed from .VGD filenames/core-level names.
    "C1s": "#8C8C8C",
    "C1s_Scan": "#8C8C8C",
    "N1s": "#2F80ED",
    "N1s_Scan": "#2F80ED",
    "O1s": "#EB5757",
    "O1s_Scan": "#EB5757",
    "S2p": "#F2C94C",
    "S2p_Scan": "#F2C94C",
    "Cl2p": "#27AE60",
    "Cl2p_Scan": "#27AE60",
}
```

## Survey Scan

```python
vgd_path = Path("PDI-H-COOH/XPS Survey.VGD")
output_dir = Path("processed_xps/PDI-H-COOH")

output_name = "pdi_h_cooh_xps_survey"
figure_title = "PDI-H-COOH XPS Survey Scan"

xps_df = read_vgd_to_dataframe(vgd_path)
xps_df = add_normalised_intensity(xps_df, intensity_column="corrected_intensity")

save_xps_outputs(
    xps_df=xps_df,
    output_dir=output_dir,
    output_name=output_name,
)

plot_normalised_xps(
    xps_df=xps_df,
    output_dir=output_dir,
    output_name=output_name,
    figure_title=figure_title,
    mode="survey",
    colour_map=CORE_LEVEL_COLOURS,
)
```

## PDI-H-COOH

```python
vgd_path = Path("PDI-H-COOH/C1s Scan.VGD")
output_dir = Path("processed_xps/PDI-H-COOH")

output_name = "pdi_h_cooh_xps_c1s_scan"
figure_title = "PDI-H-COOH C 1s Scan"

xps_df = read_vgd_to_dataframe(vgd_path)
xps_df = add_normalised_intensity(xps_df, intensity_column="corrected_intensity")

peak_df = extract_peak_maximum(
    xps_df=xps_df,
    be_min=PEAK_WINDOWS["C 1s"][0],
    be_max=PEAK_WINDOWS["C 1s"][1],
    intensity_column="corrected_intensity",
    smoothing_window=5,
)

save_xps_outputs(
    xps_df=xps_df,
    output_dir=output_dir,
    output_name=output_name,
)

save_khervefitting_xlsx(
    xps_df=xps_df,
    output_dir=output_dir,
    output_name=output_name,
    intensity_column="corrected_intensity",
)

peak_df.to_csv(output_dir / f"{output_name}_peak_maximum.csv", index=False)

plot_normalised_xps(
    xps_df=xps_df,
    output_dir=output_dir,
    output_name=output_name,
    figure_title=figure_title,
    mode="core_level",
    peak_df=peak_df,
    colour_map=CORE_LEVEL_COLOURS,
)
```
